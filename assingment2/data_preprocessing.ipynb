# House Prices Data Preprocessing
This notebook demonstrates data preprocessing steps for the House Prices dataset.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, MaxAbsScaler, StandardScaler, Normalizer
from categorical_encoding import *

# Load synthetic mock dataset
data = {
    'LotArea': [8450, 9600, 11250, 9550, 14260, np.nan],
    'YearBuilt': [2003, 1976, 2001, 1915, 2000, 1993],
    'Neighborhood': ['CollgCr', 'Veenker', 'CollgCr', 'Crawfor', 'NoRidge', 'CollgCr'],
    'HouseStyle': ['2Story', '1Story', '2Story', '2Story', '2Story', '1Story'],
    'SalePrice': [208500, 181500, 223500, 140000, 250000, 143000]
}
df = pd.DataFrame(data)

# Data Cleaning
df['LotArea'].fillna(df['LotArea'].median(), inplace=True)
df.drop_duplicates(inplace=True)
df['YearBuilt'] = df['YearBuilt'].astype(int)

# Outlier treatment (IQR)
Q1, Q3 = df['LotArea'].quantile([0.25, 0.75])
IQR = Q3 - Q1
lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
df['LotArea'] = np.where(df['LotArea'] > upper, upper, df['LotArea'])

# Encoding
df = one_hot_encode(df, 'HouseStyle')
df = label_encode(df, 'Neighborhood')
df = frequency_encode(df, 'Neighborhood')
df = target_encode(df, 'Neighborhood', 'SalePrice')

# Scaling
scalers = {
    'minmax': MinMaxScaler(),
    'maxabs': MaxAbsScaler(),
    'zscore': StandardScaler(),
    'normalize': Normalizer()
}

scaled_results = {}
for name, scaler in scalers.items():
    scaled_results[name] = pd.DataFrame(
        scaler.fit_transform(df[['LotArea', 'YearBuilt', 'SalePrice']]),
        columns=['LotArea', 'YearBuilt', 'SalePrice']
    )

# Log transformation
df['SalePrice_log'] = np.log1p(df['SalePrice'])
df.to_csv('dataset/preprocessed_mock_data.csv', index=False)
print('Preprocessing complete! Synthetic dataset processed successfully.')
```
